# Backpropagation
Short for Backward Propagation of Errors in supervised learning of ANNs using Gradient Descent.

Simply put, it is an algorithm to train artificial neural networks

___Steps:___   

0. Initialize weights and bias
1. Use forward propagation and make predictions
2. Choose a loss function and calculate losses
3. Update weights and bias using _Gradient Descent_
$$W_{new} = W_{old} - \eta \frac{\partial{L}}{\partial{W_{old}}}$$

(same for bias)

- ___Intuition behind this update function___:  


If change in L wrt $W_{old}$ is positive --> that means as W↑ L↑, so if we want L to decrease we have to decrease W --> hence subtraction

If change in L wrt $W_{old}$ is negative --> that means as W↑ L↓, so if we want L to decrease we have to increase W --> hence addition

- ___Intuition behind learning rate___

Learning rate is the size of steps in update function. If Learning rate is high then new values of weight may be very far from each other and may never converge. And if Lr is low then new values of weight will be very close and may take very high number of epochs to converge

Generally $\eta$ = 0.1 or 0.01 

- ___What is convergence?___   


Convergence in gradient descent is the point where the algorithm has minimized the cost function enough that further iterations do not significantly reduce the loss, indicating the model has found an optimal or near-optimal solution for the weights and bias

````python
#Initialization funciton - layer architecture will be given as list
def initialize_param(layer_dim: list):
    parameters = {}
    L = len(layer_dim)

    for i in range(1,L):
        parameters['W' + str(i)] = np.ones(shape=(layer_dim[i-1],layer_dim[i]))
        parameters['b' + str(i)] = np.zeros(shape=1)

    return parameters

#Test
parameters =initialize_param([2,2,1])
parameters
````

````python
def summation(X, W, b):

    z = (W.T @ X) + b
    return z

#Test
z1 = summation(df.drop(columns='lpa').iloc[0], parameters['W1'], parameters['b1'])
z2 = summation(z1, parameters['W2'],parameters['b2'])
z2
````

````python
def loss(y,y_hat):
    return (y - y_hat)**2

#Test
loss(df.iloc[0,2], z2)
````
