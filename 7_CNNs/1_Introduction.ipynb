## Convolutional Neural Networks
CNNs or Convnet, are special kind of neural network made for processing data that has grid-like topology like time-series data(1D) or Images(2D)

- How is it different from ANNs?  
There are three types of layers in CNNs: Convolutional, Pooling, and Fully Connected layer  
ANNs have only one of these layer - fully connected one.  
In layers of ANNs, matrix multiplication is performed while, in Convolutional layers of CNNs, convolution is performed.

- It is inspired from `Visual Cortex` of human brain 

- History:   
Yann Lecun, a computer scientist at AT&T lab, in 1998 built an Convolutional Network that can scan bank checks   
After that microsoft built many OCRs, hand writting recognition models using CNNs   
Now CNNs are widely accepted in many tasks such as facial recognition, Self driving cars 

Why not use ANNs?   
1. High computation:   
To give images as input to ANNs, we have to flatten the image from 2D to 1D, hence many many features. When given to network for training it takes a lot of processing power

2. Information loss   
Since we are flattening the images, information like spatial arrangement of pixels gets lost. Which may be very important in some cases like facial recognition

3. Overfitting  
Since we have so many connections in network, we may end up capturing all the noise in train images and end up with bad generalisation

### The Famous Cat Experiment
In 1960s, David Hubel and Torsten Wiesel, two neuroscientists discovered the way our brain recognises images:  
_Our brain dont recognize shapes as whole, but builds them piece by piece, starting from edges, then complex shapes, then more complex shapes_

This inspired one of the most powerful AI tool: Convolutional Neural Networks 

Our Visual Cortex:    


___The famous Cat Experiment:___

- The cat was given anesthesia, only enough to keep its eyes open and all other activties shut
- Then electrodes were place in cells of its visual cortex
- It was expected that, when neurons will react to simple dots of light. But the neurons showed no activity
- But then, by accident neuron activity fired intensly and thick black lines appeard

This proved certain points that: 
1. Our brain does not see an image as whole, it detects small edges first
2. Different neurons react to different angles and orientations 
3. Higher level neurons then combine these edge to build shapes, then even higher level neurons combines those shapes to build even more complex shape

_This meant that our brains don’t recognize objects as a whole from the start. Instead, they build them step by step, like putting together pieces of a puzzle._



Note:  
What are `edges`:   
Edges are basically boundaries in image where intensity of pixels change
