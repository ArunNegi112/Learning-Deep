## CNNs And ANNs
- How CNNs work:
    - We take input image (2D or 3D), Pass it through convolution layer
    - In convolution layer, for each filter in this layer we have a bias value (this bias value will be added to each pixel in the respective feature map). Then Activation operation is performed (ex: Relu), then output volume is passed for pooling
    - After that output is flatten into 1D Tensor and passed to dense layers 


- How ANNs work:
    - We directly flatten the images and pass it to dense layers

### The similarity between CNNs and ANNs
- In ANNs, for each neuron in any layer we get:
    $$Relu(w_1x_1 + w_2x_2 + ... + b)$$
    - Where $w_1, w_2, b$ are weights and bias


- This is similar for each filter in a convolution layer:
    - Each element in feature map is given by
    $$Relu(w_1x_1 + w_2x_2 + ... + b)$$
    - Where $w_1, w_2, b$ are elements of feature map (which are trainable parameters)


So, basically each feature map in CNNs, acts as neuron in ANNs. But difference is that CNNs can capture patterns in 2D or 3D, because of how convolution works fundamentally.

And this is ___How CNNs capture spatial arrangement of pixels___

### How CNNs reduce computation cost

_Calculate number trainable parameters_:
- CNNs: 
    Let the input be image be of (228x228x3), and we have 50 filters with shape (3x3x3), then we have output of (226x226x50).    
    How many trainable parameters are there?  
    $$ 3 \times 3 \times 3 \times 50 + 50 = 1350 + 40 = 1400$$
    Now even if input image is of (1028x1028x3) the trainable parameters will still be 1400

- ANNs:
    Let the input be image be of (228x228x3), and we have 50 neurons     
    How many trainable parameters are there?  
    $$ 228 \times 228 \times 3 \times 50 + 50 = 7797650 $$
    Huge number of trainable parameters. And this will increase with increase in input size
