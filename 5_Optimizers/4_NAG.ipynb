## Nesterov Acceleration Gradient 
Nesterov Accelerated Gradient (NAG) improves SGD with momentum by computing the gradient at a lookahead position (future estimate of parameters). This allows the optimizer to anticipate the direction of movement, reducing oscillations and improving convergence near minima.

NAG does not calculate gradient of current weight.   
It calculates graident at a lookahead position. 

- We calculate a lookahead position using velocity  
$W_{la} = W_t - \beta V_{t-1}$

- Then we calculate gradients at that position
- Then we use those gradients at lookahead position to update the velocity    
$V_{t} = \beta V_{t-1} + \eta \nabla W_{la}$  
$V_{t} = (W_t - W_{la}) + \eta \nabla W_{la}$  

- Hence finally update rules looks like this  
$W_{t+1} = W_t - V_t$   
$W_{t+1} = W_t - W_t + W_{la} - \eta \nabla W_{la}$    
$W_{t+1} = W_{la} - \eta \nabla W_{la}$

In short: 
- First we move to lookahead position using velocity
- Then we calculate gradient at this position
- Then we calculate new velocity using those gradients
- Then we update using that new velocity 

In geometry this looks like:
- We first move to a lookahead position using velocity
- Then updating the weights from that position  
$W_{t+1} = W_{la} - \eta \nabla W_{la}$
