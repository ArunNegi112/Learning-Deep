## Adam (Adaptive Moment Estimation) Optimizer
Adam combines the two most powerful optimization techniques - Momentum(NAG) and RMSProp
- Momentum(NAG): It uses ewma of gradients to update the weights 
- RMSProp: improved upon AdaGrad, this uses ewma of square of gradients to change learning rate during training.

$$
W_{t+1} = W_t - \frac{\eta}{\sqrt{V_t + \epsilon}} m_t
$$

where,   
$
m_t = \beta_1 m_{t-1} + (1-\beta_1) \nabla W_t
$

$
V_t = \beta_2 V_{t-1} + (1-\beta_2) (\nabla W_t)^2
$

generally value of
- $\beta_1$ = 0.9
- $\beta_2$ = 0.99

Then comes sparse data

Adam has benefits of All - Momentum, AdaGrad

### Verdict
- Start with `Adam` if not satisfied with results 
- Then check other optimizers - RMSProp, NAG, Momentum
