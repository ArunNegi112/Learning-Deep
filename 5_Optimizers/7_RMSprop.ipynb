## RMSProp (Root Mean Squared Propagation)
A simple upgrade on adagrad 


$$
W_{t+1} = W_t + \frac {\eta}{\sqrt{V_t + \epsilon}} \nabla W_t
$$

Where,
$$
V_t = \beta V_{t-1} + (1- \beta) (\nabla W_t)^2
$$
$$

By changing the formula of $V_t$ to ewma of squares of gradient. The learning rate now, do not get diminished so much that it wont converge. Because initial gradients are weighted eponentially smaller in the summation of $V_t$

RMSProp can be used in both convex and non-convex optimization problem
