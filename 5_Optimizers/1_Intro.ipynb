## Optimizers 
Training a neural network is an optimization problem, We start from vectors of random weights to evaluate the outputs and then change the weights to minimize the loss.

The optimizer we use is Gradient Descent

### Problem with Gradient Descent
- Learning Rate: lr becomes the most crucial factor in training time. As with time, gradients keep getting smaller but learning rate is fixed, If learning rate is small then convergence is slow, If learning rate is large, convergence may not even happen. 

- Learning rate scheduler: helps by chaning learning rate during training but the problem is that "it changes learning rate according to a predefined strategy" and not according to data itself, so one learning rate schedule may not be universal for all datasets

- Learning rate is same in all directions: Lets say we have 10 weights, then the weights space have 10 directions, and since learning rate is same for all direction, the gradient moves with same step in each direction for the minima.  
Obviously itll be better if learning rate is different for each direction,  
check the example below with 3D weights space

- Local minima: Algo may get stuck in local minima (suboptimum solution).
- Saddle point: a point where the overall gradient is zero, but the point is not a local minimum or a local maximum.
