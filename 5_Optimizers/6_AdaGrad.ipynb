## AdaGrad (Adaptive Gradient Descent)

Gradient Descent and other conventional optimization techniques use a fixed learning rate throughout the duration of training. However, this uniform learning rate might not be the best option for all parameters, which could cause problems with convergence. Some parameters might need more frequent updates to hasten convergence, while others might need smaller changes to prevent overshooting the ideal value.


_AdaGrad (Adaptive Gradient Algorithm) is one such algorithm that adjusts the learning rate for each parameter based on its prior gradients._

### Adaptive Gradient update formula
$$
W_{t+1} = W_t + \frac {\eta}{\sqrt{V_t + \epsilon}} \nabla W_t
$$

Where,
$$
V_t = V_{t-1} + (\nabla W_t)^2
$$

- $\epsilon$ is a very smalll error term, to prevent division by zero in case $V_t$ is zero
- We used $(\nabla W)^2$ so that +ve and -ve gradients do not cancel each other
- Overall learning rate now depends on summation of all past gradients
- If gradients are large, overall learning rate becomes small
- If gradient are small, overall learning rate becomes big

### Disadavantages of Adagrad
- Adagrad is generally used in linear regression and other simple algorithms and not neural network, because it does take shorter path to minima but take very high number of iterations to actually converge at minima   
This is because as iterations increase learning rate diminishes as $V_t$ becomes large

In short:   
    - Adagrad work properly and converges properly for convex optimization problems like linear regression   
    - But does not converge to gloabal minima for non-convex problems like neural network
