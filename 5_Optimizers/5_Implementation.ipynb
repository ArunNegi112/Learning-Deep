```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from keras import Sequential
from keras.optimizers import SGD
from keras.layers import Input, Dense
```

```python
#Get and visualize data
data = pd.read_csv("daily-minimum-temperatures-in-me.csv")
data.rename(columns={'Daily minimum temperatures': 'Temperature'}, inplace=True)
data['Date'] = pd.to_datetime(data['Date'])
plt.scatter(data['Date'],data['Temperature'],s=5)
plt.xlabel("Date")
plt.ylabel("Temperature")
```

```python
#Divide features and target
X = pd.concat(
    [data['Date'].dt.day, data['Date'].dt.month], 
    axis=1
    )
y = data['Temperature']
```

```python
#Structure the model
model = Sequential()
model.add(Input((X.shape[1],)))
model.add(Dense(16, activation='relu'))

#SGD
sgd = SGD(
    learning_rate=0.01,
    momentum=0.0,
    nesterov=False
)
model.compile(optimizer=sgd, loss='mse', metrics=['root_mean_squared_error'])
history_sgd = model.fit(X,y, epochs=100, verbose=1)
```

```python
model = Sequential()
model.add(Input((X.shape[1],)))
model.add(Dense(16, activation='relu'))

#SGD with momentum
momentum = SGD(
    learning_rate = 0.01,
    momentum = 0.9, #value of beta
    nesterov=False
)

model.compile(optimizer=momentum, loss='mse', metrics=['root_mean_squared_error'])
history_momentum = model.fit(X,y, epochs=100, verbose=1)
```

```python
plt.plot(history_momentum.history['root_mean_squared_error'])
```

```python
model = Sequential()
model.add(Input((X.shape[1],)))
model.add(Dense(16, activation='relu'))

#NAG
nag = SGD(
    learning_rate = 0.01,
    momentum = 0.9, #value of beta
    nesterov=True
)

model.compile(optimizer=nag, loss='mse', metrics=['root_mean_squared_error'])
history_nag = model.fit(X,y, epochs=100, verbose=1)
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.plot(history_sgd.history['root_mean_squared_error'])
plt.title('SGD')

plt.subplot(1,3,2)
plt.plot(history_momentum.history['root_mean_squared_error'])
plt.title('SGD with Momentum')

plt.subplot(1,3,3)
plt.plot(history_nag.history['root_mean_squared_error'])
plt.title('NAG')

plt.show()
```
