### 1. Fine tuning Hyperparameters


1. __Number of hidden layers__
- We can experiment with number of layers until you get overfitting
- _More neurons and less layers_ OR _less neurons and more layer_? 
    - Neural Networks have Hierarchical feature learning
        - Means, initial layers learn simple patterns and later layers combine them into complex patters


2. __How many Number of neurons?__
- Sufficient.
- Number of neurons should be sufficient to capture all the patterns in data, if overfitting or high training time is observed then reduce as required

6. __Batch size__  
Highly important btw, as it controls training time and performance both

Try to keep batch_size as $2^n$ 
- Small batch size (2,4,8,16,32,64,128,256)
    - Slow
    - Generalize better 


- Large batch size (512,1024,2048,4096,8192)
    - Fast
    - Fails in generalization. i.e Perform good in training data but fails on unseen data


___Note___:  
Hence, Try to experiment with big batch size, and if it performs well on validation set, then it is better (because of less training time).  
We can also achieve this by varying _learning_rate_ wrt epochs, 
    
- _Learning rate scheduler_ : Initially keep learning rate small, in later iterations increase the learning rate gradually. This is called `warming up the learning rate`


_IF these techniques do not work then go back to small batch_size, that will take time but definitely give results_
